```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

# Dataset path
DATA_DIR = Path("C:/Users/HP/Desktop/New folder/ML junior/PlantDiseaseClassification/data/color")

# Class names, Picture numbers in each class, Number of classes
num_classes = 0
total = 0
class_names = []
image_counts = []
print("subfolder names:")
for folder in DATA_DIR.iterdir():
    if folder.is_dir():
        num_classes = num_classes + 1
        image_count = 0
        extensions = ["*.jpg", "*.jpeg", "*.png"]
        for extension in extensions:
            image_count = image_count + len(list(folder.glob(extension)))
        total = total + (image_count)
        class_names.append(folder.name)
        image_counts.append(image_count)
        print(folder.name, ":", image_count, "pictures")
print("\n\n\nnumber of classes : ", num_classes)
print(f"total pictures count: {total}")

# dataset visualization
fig, ax = plt.subplots(figsize = (12, 16))
ax.barh(class_names, image_counts)
ax.set_ylabel("class_names")
ax.set_xlabel("number of images")
ax.set_title("PlantVillage Dataset - Class Distribution")
ax.tick_params(axis = "y", labelsize=8)
ax.invert_yaxis()
plt.tight_layout()
plt.show()

# Selecting 6 different classes
selected_classes = [
    "Apple___Apple_scab",
    "Apple___healthy",
    "Corn_(maize)___Common_rust_",
    "Grape___Black_rot",
    "Potato___Early_blight",
    "Tomato___healthy"
]

# a figure with 2 rows and 3 columns
fig, axes = plt.subplots(2, 3, figsize=(12, 8))

for ax, class_name in zip(axes.flat, selected_classes):

    # Find the selected class folder
    class_dir = DATA_DIR / class_name

    # Find image files
    image_paths = (
        list(class_dir.glob("*.jpg"))
        + list(class_dir.glob("*.jpeg"))
        + list(class_dir.glob("*.png"))
    )

    # Select the first image
    image_path = image_paths[0]

    # Open the image
    image = Image.open(image_path)

    # Display the image
    ax.imshow(image)
    ax.set_title(class_name, fontsize=8)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
